```python
import cv2
import os

path = "D:\\CV codes and projects"
img_format = ".jpeg"

image_name1 = "download"
image_name2 = "download2"

image1_path = os.path.join(path, image_name1 + img_format)
image2_path = os.path.join(path, image_name2 + img_format)

output_path = "D:\\CV codes and projects"
output_filename = "matched.jpeg"  
output_filepath = os.path.join(output_path, output_filename)

gray1 = cv2.imread(image1_path, cv2.IMREAD_GRAYSCALE)
gray2 = cv2.imread(image2_path, cv2.IMREAD_GRAYSCALE)

if gray1 is None or gray2 is None:
    raise ValueError("Failed to load images. Check paths and filenames.")

sift = cv2.SIFT_create()
keypoints1, descriptors1 = sift.detectAndCompute(gray1, None)
keypoints2, descriptors2 = sift.detectAndCompute(gray2, None)

bf = cv2.BFMatcher(cv2.NORM_L2, crossCheck=True)
matches = bf.match(descriptors1, descriptors2)
matches = sorted(matches, key=lambda x: x.distance)

matched_img = cv2.drawMatches(
    gray1, keypoints1,
    gray2, keypoints2,
    matches[:50],
    None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

cv2.imwrite(output_filepath, matched_img)
print(f"Matched image saved to: {output_filepath}")
```

```python
import cv2
import numpy as np

image_path = "D:\\CV codes and projects\\download.jpeg"
img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)  

if img is None:
    raise ValueError("Failed to load image. Check the path.")

height, width = img.shape
zoom_factor = 2.0

new_height, new_width = int(height/zoom_factor), int(width/zoom_factor)
start_x, start_y = int((width - new_width)/2), int((height - new_height)/2)
zoomed_img = img[start_y:start_y+new_height, start_x:start_x+new_width]
zoomed_img = cv2.resize(zoomed_img, (width, height)) 

sift = cv2.SIFT_create()

kp1, des1 = sift.detectAndCompute(img, None)
kp2, des2 = sift.detectAndCompute(zoomed_img, None)

FLANN_INDEX_KDTREE = 1
index_params = dict(algorithm=FLANN_INDEX_KDTREE, trees=5)
search_params = dict(checks=50)
flann = cv2.FlannBasedMatcher(index_params, search_params)

matches = flann.knnMatch(des1, des2, k=2)

good_matches = []
for m,n in matches:
    if m.distance < 0.7*n.distance:
        good_matches.append(m)

matched_img = cv2.drawMatches(
    img, kp1,
    zoomed_img, kp2,
    good_matches[:50],  
    None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

output_path = "D:\\CV codes and projects\\matched_zoom.jpeg"
cv2.imwrite(output_path, matched_img)

cv2.imshow("SIFT Matching: Original vs Zoomed", matched_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Load the image
image_path = "D:\\CV codes and projects\\picture.jpg"
img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if img is None:
    raise ValueError("Image failed to load. Please check the path.")

# Create flipped version (upside down)
flipped_img = cv2.flip(img, 0)  # 0 means vertical flip

# Initialize SIFT detector
sift = cv2.SIFT_create()

# Detect keypoints and descriptors
kp1, des1 = sift.detectAndCompute(img, None)
kp2, des2 = sift.detectAndCompute(flipped_img, None)

# Feature matching using FLANN
FLANN_INDEX_KDTREE = 1
index_params = dict(algorithm=FLANN_INDEX_KDTREE, trees=5)
search_params = dict(checks=50)
flann = cv2.FlannBasedMatcher(index_params, search_params)

matches = flann.knnMatch(des1, des2, k=2)

# Apply Lowe's ratio test
good_matches = []
for m, n in matches:
    if m.distance < 0.7 * n.distance:
        good_matches.append(m)

# Draw matches
matched_img = cv2.drawMatches(
    img, kp1,
    flipped_img, kp2,
    good_matches[:50],  # Show top 50 matches
    None,
    matchColor=(0, 255, 0),  # Green lines
    singlePointColor=(255, 0, 0),  # Red points
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

# Save result
output_path = "D:\\CV codes and projects\\matched_flipped.jpg"
cv2.imwrite(output_path, matched_img)

# Display result
cv2.imshow("SIFT Matching: Original vs Flipped", matched_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
